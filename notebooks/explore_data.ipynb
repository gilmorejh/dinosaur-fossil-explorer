```python
column_profile = pd.DataFrame({
    "column": df.columns,
    "dtype": df.dtypes.astype(str),
    "missing": df.isna().sum(),
    "missing_pct": (df.isna().mean() * 100).round(2),
    "unique": df.nunique()
})

column_profile
```

```python
column_profile.sort_values("missing_pct", ascending=False).head(30)
```

```python
column_profile.sort_values("missing_pct").head(30)
```

```python
column_profile[
    (column_profile["missing_pct"] >= 20) &
    (column_profile["missing_pct"] <= 80)
].sort_values("missing_pct")
```
