```python
import requests
import pandas
import json
import os
from pathlib import Path
```

```python
#Test endpoint to get a sample of dinosaur occurrence data from the Paleobiology Database API
occurrence_url = "https://paleobiodb.org/data1.2/occs/list.json?base_name=Dinosauria&show=full&limit=10"

response = requests.get(occurrence_url)
response.raise_for_status()

occurrence_data = response.json()

occurrence_df = pandas.DataFrame(occurrence_data["records"])

occurrence_df.head()
```

```python
#Retrieve the full dataset of dinosaur occurrences from the Paleobiology Database API
full_url = "https://paleobiodb.org/data1.2/occs/list.json?base_name=Dinosauria&show=full&limit=50000"
response = requests.get(full_url)
response.raise_for_status()

full_data = response.json()
full_df = pandas.DataFrame(full_data['records'])
```

```python
#Test length of the full dataset and check for duplicates
print(len(full_df))
print(len(full_data['records']))
print(full_df['oid'].nunique())
print(full_df['oid'].duplicated().sum())
```
